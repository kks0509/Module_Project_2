1번 학습용 DoS 수집 데이터

vmware에서 slowloris 공격하여 얻은 다량의 패킷 데이터(pcap) 파일에 cicflowmeter 적용하여 csv로 변환

input 데이터
1. slowdata.pcap : vmware에서 slowloris 공격하여 얻은 다량의 패킷 데이터

output 데이터
1. slowdata.csv : input 데이터(pcap) 파일을 csv 형식으로 변환

```python
# pcap에 cicflowmeter 적용 후 csv 파일 변환
import sys
import scapy.sendrecv as sr

sr.tcpdump = lambda fname, *args, **kwargs: open(fname, "rb")

from cicflowmeter.sniffer import main

pcap_path = r"C:\python_proj\모듈 프로젝트2\cicflowmeter\교육용 수집 데이터\slowdata.pcap"
csv_path = r"C:\python_proj\모듈 프로젝트2\cicflowmeter\교육용 수집 데이터\slowdata.csv"

sys.argv = ["cicflowmeter", "-f", pcap_path, "-c", csv_path]
main()
```
